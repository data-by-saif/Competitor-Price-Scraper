```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import re

# Website URL
base_url = "https://books.toscrape.com/catalogue/page-1.html"

print("Opening website...")

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(base_url, headers=headers)

if response.status_code != 200:
    print(f"Could not access the website. Status code: {response.status_code}")
    exit()

print("Website loaded successfully.")

# Parse HTML
soup = BeautifulSoup(response.text, "html.parser")

books = []

# Find all book cards
for book in soup.find_all("article", class_="product_pod"):

    # Title
    title = book.h3.a["title"]

    # Price
    price_text = book.find("p", class_="price_color").text

    # Remove currency symbols and unwanted characters
    price = float(re.sub(r"[^\d.]", "", price_text))

    # Availability
    availability = book.find(
        "p",
        class_="instock availability"
    ).text.strip()

    # Product URL
    link = book.h3.a["href"]
    product_url = f"https://books.toscrape.com/catalogue/{link}"

    books.append({
        "Title": title,
        "Price_GBP": price,
        "Availability": availability,
        "Product_URL": product_url
    })

# Create DataFrame
df = pd.DataFrame(books)

# Save CSV
file_name = "books_catalog.csv"
df.to_csv(file_name, index=False)

print(f"\nSaved {len(df)} books to '{file_name}'")

# Display first 10 rows
print("\nFirst 10 books:")
print(df.head(10))
```
